```python
import sys
import shutil
import subprocess
import tempfile
import os

print("🏋️‍♀️ Ms. Olympia's Toolchain Inspection — Finding Our Best Friends in the Bunker!\n")
print(f"Active Notebook Python : {sys.executable} (Python {sys.version.split()[0]})")
for tool in ["dotnet", "uv", "winget", "powershell"]:
    print(f"Found Host Toolchain   : {tool:10s} -> {shutil.which(tool)}")

# Demonstrate F# 10.0 (.NET 10 SDK) — Microsoft's ML-family cousin to Rust — solving Eugen Valuev natively!
fsharp_code = r"""
open System
open System.Numerics

// In F# (like Rust): immutable by default, algebraic types, strict type inference, batteries-included BCL!
type BiomeStatus =
    | StableDetente of eigenvalues: float array
    | RunawayResonance of dominant: float

// Pure F# Symmetric Jacobi Eigenvalue Solver (zero external dependencies!)
let jacobiEigenvalues (aInput: float[,]) (maxSweeps: int) =
    let n = Array2D.length1 aInput
    let a = Array2D.copy aInput
    for _ in 1 .. maxSweeps do
        for p in 0 .. n - 2 do
            for q in p + 1 .. n - 1 do
                let apq = a.[p, q]
                if abs apq > 1e-12 then
                    let app = a.[p, p]
                    let aqq = a.[q, q]
                    let tau = (aqq - app) / (2.0 * apq)
                    let t =
                        if tau >= 0.0 then 1.0 / (tau + sqrt(1.0 + tau * tau))
                        else -1.0 / (-tau + sqrt(1.0 + tau * tau))
                    let c = 1.0 / sqrt(1.0 + t * t)
                    let s = t * c
                    a.[p, p] <- app - t * apq
                    a.[q, q] <- aqq + t * apq
                    a.[p, q] <- 0.0
                    a.[q, p] <- 0.0
                    for r in 0 .. n - 1 do
                        if r <> p && r <> q then
                            let arp = a.[r, p]
                            let arq = a.[r, q]
                            a.[r, p] <- c * arp - s * arq
                            a.[p, r] <- a.[r, p]
                            a.[r, q] <- s * arp + c * arq
                            a.[q, r] <- a.[r, q]
    [| for i in 0 .. n - 1 -> a.[i, i] |] |> Array.sortDescending

// Spawn a 6x6 Wigner Symmetric Barbell Matrix in F# 10
let rng = Random(1989)
let n = 6
let wigner = Array2D.init n n (fun i j ->
    if i = j then (rng.NextDouble() - 0.5) * 2.0
    elif i < j then (rng.NextDouble() - 0.5) * 2.0
    else 0.0)
for i in 0 .. n - 1 do
    for j in 0 .. i - 1 do
        wigner.[i, j] <- wigner.[j, i]

let eugens = jacobiEigenvalues wigner 30
printfn "\n[F# 10.0 / .NET 10.0.403] Zero-Dependency Jacobi Solver Online!"
printfn "Emergent Eugen Valuevs (lambda_1 .. lambda_6): %s" (String.Join(", ", eugens |> Array.map (fun x -> sprintf "%+.4f" x)))
printfn "Spectral Radius rho(A) = %.4f | Spectral Gap d_lambda = %.4f" (eugens |> Array.map abs |> Array.max) (abs eugens.[0] - abs eugens.[1])
"""

with tempfile.NamedTemporaryFile(mode="w", suffix=".fsx", delete=False, encoding="utf-8") as f:
    f.write(fsharp_code)
    fsx_path = f.name

try:
    res = subprocess.run(
        ["dotnet", "fsi", fsx_path],
        capture_output=True,
        text=True,
        encoding="utf-8",
        errors="replace",
        check=True,
    )
    print(res.stdout)
finally:
    os.remove(fsx_path)
```
